```python
#question 3
df = pd.DataFrame(np.arange(20).reshape(-1, 5), columns=list('abcde'))
def swap_columns(df,column1,column2):
    columns = list(df.columns)
    place1 = columns.index(column1)
    place2 = columns.index(column2)
    columns[place1], columns[place2] = columns[place2], columns[place1]
    return df[columns]
swap_df = swap_columns(df, 'b', 'e')
swap_df
```
